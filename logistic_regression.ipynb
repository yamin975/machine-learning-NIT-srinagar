# **`logistic with Confusion Matrix`**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

from sklearn.metrics import accuracy_score
from sklearn.metrics import  confusion_matrix
from sklearn.metrics import   ConfusionMatrixDisplay
from sklearn.metrics import  precision_score
from sklearn.metrics import  recall_score
from sklearn.metrics import  f1_score

from sklearn.preprocessing import StandardScaler
```

libraries


```python
b_data=load_breast_cancer()
xt4=b_data.data
yt9=b_data.target
train, test, y_train, y_test=train_test_split(xt4,yt9,test_size=0.2, random_state=101, stratify=yt9)
model=LogisticRegression(max_iter=1000)
model.fit(train,y_train)
y_pred=model.predict(test)


accuracy=accuracy_score(y_test,y_pred)
precision=precision_score(y_test,y_pred)
recall=recall_score(y_test,y_pred)
f1=f1_score(y_test,y_pred)


print(f"accuracy is: {accuracy}")
print(f"precision is: {precision}")
print(f"recall is: {recall}")
print(f"f1 is: {f1}")
```

```python
cm=confusion_matrix(y_test,y_pred)
disp=ConfusionMatrixDisplay(confusion_matrix=cm,display_labels=b_data.target_names)
disp.plot()
plt.show()
```


# ***`Ridge through confusion matrix`***\

```python
ridge_model=LogisticRegression(penalty='l2',C=0.1,max_iter=1000)
ridge_model.fit(train,y_train)
y_pred_ridge=ridge_model.predict(test)

accuracy_ridge=accuracy_score(y_test,y_pred_ridge)
precision_ridge=precision_score(y_test,y_pred_ridge,pos_label=0)
recall_ridge=recall_score(y_test,y_pred_ridge,pos_label=0)
f1_ridge=f1_score(y_test,y_pred_ridge,pos_label=0)


print(f"accuracy is: {accuracy_ridge}")
print(f"precision is: {precision_ridge}")
print(f"recall is: {recall_ridge}")
print(f"f1 is: {f1_ridge}")
```

```python
rd=ridge_model.predict(test)
cm=confusion_matrix(y_test,rd)
disp=ConfusionMatrixDisplay(confusion_matrix=cm,display_labels=b_data.target_names)
disp.plot()
plt.show()
```

# ***`Ridge through confusion matrix`***

```python

lasso_model=LogisticRegression(penalty='l1',C=0.1,max_iter=1000,solver='liblinear')
lasso_model.fit(train,y_train)
y_pred_lasso=lasso_model.predict(test)

accuracy_lasso=accuracy_score(y_test,y_pred_lasso)
precision_lasso=precision_score(y_test,y_pred_lasso,pos_label=0)
recall_lasso=recall_score(y_test,y_pred_lasso,pos_label=0)
f1_lasso=f1_score(y_test,y_pred_lasso,pos_label=0)


print(f"accuracy is: {accuracy_lasso}")
print(f"precision is: {precision_lasso}")
print(f"recall is: {recall_lasso}")
print(f"f1 is: {f1_lasso}")
```

```python
lasso_model.predict(test)
cm=confusion_matrix(y_test,y_pred_lasso)
disp=ConfusionMatrixDisplay(confusion_matrix=cm,display_labels=b_data.target_names)
disp.plot()
plt.show()
```

# ***`Ridge through confusion matrix`***


```python
elastic_model=LogisticRegression(penalty='elasticnet',
                                 l1_ratio=0.5,
                                 solver='saga',
                                 C=1,
                                 max_iter=5000)
elastic_model.fit(train, y_train)
y_pred_elastic=elastic_model.predict(test)



accuracy_net=accuracy_score(y_test,y_pred_elastic)
precision_net=precision_score(y_test,y_pred_elastic,pos_label=0)
recall_net=recall_score(y_test,y_pred_elastic,pos_label=0)
f1_net=f1_score(y_test,y_pred_elastic,pos_label=0)


print(f"accuracy is: {accuracy_net}")
print(f"precision is: {precision_net}")
print(f"recall is: {recall_net}")
print(f"f1 is: {f1_net}")
```

```python
c4=confusion_matrix(y_test,y_pred_elastic)
disp=ConfusionMatrixDisplay(confusion_matrix=c4,display_labels=b_data.target_names)
disp.plot()
plt.show()
```

```python
compiled_results=pd.DataFrame({
    'model':[
        'logistic',
        'ridge',
        'lasso',
        'elastic net'
    ],
    'accuracy':[
        accuracy,
        accuracy_ridge,
        accuracy_lasso,
        accuracy_net
    ],
    'precision':[
        precision,
        precision_ridge,
        precision_lasso,
        precision_net
    ],
    'recall':[
        recall,
        recall_ridge,
        recall_lasso,
        recall_net
    ],
    'f1':[
        f1,
        f1_ridge,
        f1_lasso,
        f1_net
    ]
})
print(compiled_results)
```
