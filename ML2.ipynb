```python
y_true = [3.0, -0.5, 2.0, 7.0]
y_pred = [2.5, 0.0, 2.0, 8.0]

# Calculate squared differences, sum them up, and divide by total elements
mse = sum((true - pred) ** 2 for true, pred in zip(y_true, y_pred)) / len(y_true)
print(f"MSE: {mse}")  # Output: 0.375
```

```python
y_true = [3, 5, 2.0, 7.0]
y_pred = [2.5, 0.0, 2.0, 8.0]

total_error = 0

for i in range(len(y_pred)):
    # Access individual elements using the index [i]
    error = y_pred[i] - y_true[i]
    total_error = total_error + error

print(f"Total Error: {total_error}")  # Output: -4.5
```
