# ***`Logistic Regression Decision Boundary`***

```python
import numpy as np
import matplotlib.pyplot as plt


from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression

data_b= load_breast_cancer()
x44=data_b.data
y99=data_b.target


x44=x44[:,[0,1]]

print("features:", data_b.feature_names[[0,1]])
print("x44 shape", x44.shape
      )
```

```python
train, test, y_train, y_test= train_test_split(
    x44,y99,test_size=0.2,random_state=101,stratify=y99
)

model_ansar=LogisticRegression(
    penalty='l2',
    solver='lbfgs',
    max_iter=1000
)
model_ansar.fit(train,y_train)
print("score:", model_ansar.score(test,y_test))
```

```python
theta_0 = model_ansar.intercept_[0]
theta_1 = model_ansar.coef_[0][0]
theta_2 = model_ansar.coef_[0][1]

print("theta_0 =", theta_0)
print("theta_1 =", theta_1)
print("theta_2 =", theta_2)
```

```python
x1_values = np.linspace(
    x44[:, 0].min(),
    x44[:, 0].max(),
    100
)

x2_values = -(theta_0 + theta_1 * x1_values) / theta_2
print(x2_values)
```

```python
plt.figure(figsize=(6, 6))

plt.scatter(
    x44[y99 == 0, 0],
    x44[y99 == 0, 1],
    label='Malignant'
)

plt.scatter(
    x44[y99 == 1, 0],
    x44[y99 == 1, 1],
    label='Benign'
)
plt.plot(
    x1_values,
    x2_values,
    linewidth=2,
    label='Decision Boundary'
)

plt.xlabel('Mean Radius')
plt.ylabel('Mean Texture')
plt.title('Logistic Regression Decision Boundary')

plt.legend()
plt.grid()
plt.show()
```
